# Data Cleaning

Before performing exploratory data analysis, the dataset is cleaned to ensure consistency, accuracy, and readability.

The cleaning process includes:

1. Renaming columns for clarity  
2. Dropping irrelevant columns  
3. Fixing column text and data types  
4. Handling missing values  
5. Preparing columns for analysis and visualization

### Step 1: Rename Columns

- Column names are simplified for readability and easier referencing in code.  
- **Example:** `Product_Category`, `Sales_Volume`, `Product_Position` have readable and consistent names.


```python
df.rename(columns={
    'Product_ID': 'Product_ID',
    'Product_Position': 'Product_Position',
    'Product_Category': 'Product_Category',
    'Sales_Volume': 'Sales',
    'Price': 'Price',
    'Promotion': 'Promotion',
    'Seasonal': 'Seasonal',
    'Name': 'Name'
}, inplace=True)

df.columns
```

### Step 2: Drop Irrelevant Columns

- Columns not needed for this analysis are removed to streamline the dataset.  
- **Example:** Dropping `brand`, `url`, `sku`, `description` since they don’t contribute to sales insights.

```python
df = df[['Product_Category', 'Sales', 'price',
         'Promotion', 'Seasonal', 'Product_Position', 'name']]
print(df.head())
```

```python
df.info()
```

### Step 3: Fix Column Text and Data Types

- Ensuring numeric and categorical columns have the correct data types and consistent text formatting.  
- **Example:** Converting `Sales_Volume` and `Price` to numeric, and `Promotion` or `Product_Position` to categorical types.

- Verify the dataset is clean and ready for analysis using:
  - `df.head()` for a quick preview  
  - `df.info()` to check rows, columns, and data types  
  - `df.describe()` for summary statistics

```python
df['Sales'] = pd.to_numeric(df['Sales'], errors='coerce')

if df['price'].dtype == 'object':
    df['price'] = pd.to_numeric(df['price'].str.replace('£', '', regex=False).str.replace('$', '', regex=False), errors='coerce')
else:
    df['price'] = pd.to_numeric(df['price'], errors='coerce')

categorical_cols = ['Promotion', 'Seasonal', 'Product_Position', 'Product_Category']
for col in categorical_cols:
    df[col] = df[col].astype('category')

df.info()
```

```python
df.head()
```

```python
df.describe()
```

### Step 4: Handle Missing Values

- Identify missing or null values in critical columns.  
- **Example:** Removing the rows where `Sales_Volume` or `Price` are missing to avoid skewing the results.

```python
df = df.dropna(subset=['Sales', 'price'])
print(df[['Sales', 'price']].isna().sum())
```

### Step 5: Prepare Columns for Analysis and Visualization

- Standardize text and unify categories to make the dataset ready for plotting.  
- **Example:** Standardizing `Product_Position` entries to avoid duplicates like “aisle” vs “Aisle”.

```python
df['Product_Position'] = df['Product_Position'].str.title().str.strip()
df['Promotion'] = df['Promotion'].str.title().str.strip()

df.head()
```

### Step 6: Create Sub-Category Column

- The original `Product_Category` column is not as informative (everything is “Clothing”), so a generated `Sub_Category` column was made based on product names.  

- **Example Sub-Categories:** Jackets, Shirts, Sweaters, Pants, Dresses/Skirts, Suits/Blazers

```python
def categorize_product(name):
    name = str(name).upper()
    if "JACKET" in name or "BOMBER" in name or "PUFFER" in name:
        return "Jacket"
    elif "SHIRT" in name or "POLO" in name or "T-SHIRT" in name:
        return "Shirt"
    elif "SWEATER" in name or "KNIT" in name or "CARDIGAN" in name:
        return "Sweater"
    elif "PANTS" in name or "JEANS" in name or "TROUSERS" in name:
        return "Pants"
    elif "DRESS" in name or "SKIRT" in name:
        return "Dress/Skirt"
    elif "SUIT" in name or "BLAZER" in name:
        return "Suit/Blazer"
    else:
        return "Other"

df['Sub_Category'] = df['name'].apply(categorize_product)

print(df[['name', 'Sub_Category']].head())
```

### Step 7: Drop `name` Column

- The `name` column contains specific product names, which are too granular for this analysis. Since there is now have a `Sub_Category` column, `name` is redundant. Dropping it will simplify the dataset and make the analysis easier.

```python
df = df.drop(columns=['name'])

print(df.head())
```

# Exploratory Data Analysis (EDA) — Product Placement for Higher Sales

Question 1: Which product placements lead to higher sales?

Focusing on:

- Total and average sales by product placement and whether products on aisles, end-caps, or front-of-store locations sell better.
- Records per product placement to see how products are distributed across different store locations.

These tables provide a baseline view of how product placement influences sales before moving into deeper analysis and visualizations.

```python
placement_sales_summary = df.groupby('Product_Position')['Sales'].agg(['sum', 'mean']).reset_index()
placement_sales_summary = placement_sales_summary.rename(columns={'sum': 'Total_Sales', 'mean': 'Average_Sales'})

print("Total and Average Sales by Product Placement:")
print(placement_sales_summary)

placement_counts = df['Product_Position'].value_counts().reset_index()
placement_counts.columns = ['Product_Position', 'Record_Count']

print("\nNumber of Records per Product Placement:")
print(placement_counts)
```

#### Product Placement Analysis

Summary of key trends in Zara's sales, looking at total and average sales by placement, as well as product distribution across store locations.

Key insights include:

1. Aisle products generate the highest total sales (177,396), primarily because there are more products located in aisles (97 records).
2. Front Of Store products have the highest average sales per product (1,873), even though there are fewer items (69 records), indicating that prime placement boosts individual product performance.
3. End-Cap placements perform moderately in both total and average sales (152,930 total; 1,778 average), serving as a balance between volume and visibility.
4. Overall, product placement significantly affects sales. More products on aisles maximize total revenue, while Front Of Store placement maximizes sales per item.

# Exploratory Data Analysis (EDA) — Sale Drivers Summary

Question 2: Which products and categories drive the most sales?

Focusing on:

- Total and average sales by sub-category
- Identify which types of products (jackets, shirts, etc.) perform best.
- Records per sub-category, how many products were promoted versus non-promoted.

```python
print("Total and Average Sales by Sub-Category:")
subcategory_sales = (
    df.groupby('Sub_Category')['Sales']
      .agg(Total_Sales='sum', Average_Sales='mean')
      .reset_index()
)
subcategory_sales['Average_Sales'] = subcategory_sales['Average_Sales'].round(2)
print(subcategory_sales.sort_values(by='Total_Sales', ascending=False))

print("\nNumber of Records per Sub-Category:")
subcategory_counts = (
    df['Sub_Category']
      .value_counts()
      .reset_index()
      .rename(columns={'index':'Sub_Category', 'Sub_Category':'Record_Count'})
)
print(subcategory_counts)

print("\nPromotion Distribution by Sub-Category:")
promotion_counts = (
    df.groupby(['Sub_Category', 'Promotion'])
      .size()
      .reset_index(name='Count')
)
print(promotion_counts)
```

####Product and Category Sales Analysis

Summary of total and average sales by sub-category, number of products, and promotion distribution.

Key insights include:

1. Jackets are the top-performing sub-category, generating the highest total sales (222,848) and representing the largest number of products (121 records).
2. Shirts and sweaters also contribute significantly to total sales (92,102 and 72,505), with average sales per product ranging from 1,771 to 1,812, showing solid performance despite fewer items than jackets.
3. Other products and pants have fewer records and lower total sales, but some “Other” items achieve the highest average sales per product (1,891), indicating that certain niche items sell well individually. (Could be due to seasons)
4. Sales are fairly balanced between promoted and non-promoted items within each sub-category. Jackets, shirts, and sweaters show nearly equal counts of promoted vs. non-promoted products, suggesting promotions are applied across categories rather than concentrated in one.

# Exploratory Data Analysis (EDA) — Pricing Summary

Question 3: How does pricing impact sales?

Focusing on:

- Average sales by price range and understading whether higher or lower-priced products sell better.
- Price distribution across sub-categories help identify which products fall into which price brackets.


```python
bins = [0, 50, 100, 150, 200, float('inf')]
labels = ['0-50', '51-100', '101-150', '151-200', '200+']

#Price Range was added for price bins
df['Price_Range'] = pd.cut(df['price'], bins=bins, labels=labels, right=True)

price_summary = df.groupby('Price_Range', observed=True)['Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean'
).reset_index()

price_summary['Average_Sales'] = price_summary['Average_Sales'].round(2)

print("Sales Summary by Price Range:")
print(price_summary)
```

#### Pricing Analysis

Summary looking into the total and average sales across different price ranges.

Key insights include:

1. Products priced 51-100 generate the highest total sales (200,553) and also the highest average sales per item (1,874), suggesting this is the sweet spot for customers.
2. Lower-priced items (0-50) also sell well overall (127,861 total), with solid average sales (1,801), showing a strong volume for budget-friendly products.
3. Mid-to-high priced items (101-150 and 151-200) see declining total and average sales, showing that fewer customers purchase more expensive items.
4. Products over 200+ have the lowest total (6,058) and average sales (1,515), suggesting very high-priced items have limited demand.
5. Overall, moderately priced products drive the most sales, while very high-priced items sell less frequently.

# Exploratory Data Analysis (EDA) — Promotions Summary

Question 4: Do promotions boost sales?

Focusing on:

- Average sales for promoted vs non-promoted products can tell whether or not promotions increase sales.
- Records by promotion status can show how many products were promoted versus non-promoted.

```python
promo_summary = df.groupby('Promotion')['Sales'].agg(
    Total_Sales='sum',
    Average_Sales='mean',
    Record_Count='count'
).reset_index()

promo_summary['Average_Sales'] = promo_summary['Average_Sales'].round(2)

print("Sales Summary by Promotion Status:")
print(promo_summary)
```

####Promotions Analysis

Summary of the total and average sales for promoted vs non-promoted products, as well as the distribution of promoted items.

Key insights include:

1. Non-promoted products generate slightly higher total sales (240,312) than promoted products (219,261), largely because there are more non-promoted items (132 records vs 120).
2. Promoted products have a slightly higher average sales per item (1,827) compared to non-promoted items (1,820), suggesting promotions may boost individual product performance.
3. The difference is relatively small, indicating that while promotions can help, the effect might be limited or depend on other factors like product type or placement.

# Exploratory Data Analysis (EDA) — Seasonal Summary

Question 5: How does seasonality affect product performance?

Focusing on:

- Average sales for seasonal vs non-seasonal products and if seasonal products sell differently.
- Records by seasonality to check how many products are marked as seasonal.
- Showing which products based on catergory sell more or less during seasonal times


```python
print("Sales Summary by Seasonality:")

seasonal_summary = (
    df.groupby('Seasonal', observed=True)['Sales']
      .agg(Total_Sales='sum', Average_Sales='mean', Record_Count='count')
      .reset_index()
)
seasonal_summary['Average_Sales'] = seasonal_summary['Average_Sales'].round(2)

print(seasonal_summary)
```

```python
#Seasonal Sales By Sub_Category
```

```python
seasonal_subcat_summary = (
    df.groupby(['Sub_Category', 'Seasonal'], observed=True)['Sales']
      .agg(Total_Sales='sum', Average_Sales='mean', Record_Count='count')
      .reset_index()
)
seasonal_subcat_summary['Average_Sales'] = seasonal_subcat_summary['Average_Sales'].round(2)

print("Sales Summary by Sub-Category and Seasonality:")
print(seasonal_subcat_summary)
```

#### Seasonal Sales Analysis by Product Type

Summary looking into total and average sales by seasonality, broken down by product type.

Key insights include:

4. Overall, the impact of seasonality varies by product type: some categories benefit from seasonal promotion (Shirts, Sweaters), while others are less dependent on seasonal trends (Jackets, Pants).

#### Insight #3

The number of records is proportional to sales: Aisle has 97 transactions, End-Cap 86, and Front Of Store 69, suggesting product placement affects transaction volume.

#Total Sales By Promotion Status

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import cumsum
from bokeh.io import output_notebook
import pandas as pd
import math

output_notebook()

promo_summary['Total_Percent'] = promo_summary['Total_Sales'] / promo_summary['Total_Sales'].sum() * 100
promo_summary['angle'] = promo_summary['Total_Sales'] / promo_summary['Total_Sales'].sum() * 2 * math.pi
promo_summary['color'] = ["#605046", "#A4968A"]

promo_summary['label'] = promo_summary.apply(
    lambda row: f"{row['Promotion']}\nTotal: {row['Total_Sales']:,}\nAvg: {row['Average_Sales']:.2f}\n{row['Total_Percent']:.1f}%", axis=1
)

source = ColumnDataSource(promo_summary)

p = figure(
    height=400, width=400,
    title="Total Sales by Promotion Status",
    toolbar_location=None,
    tools="hover, tap",
    tooltips="@label"
)

p.annular_wedge(
    x=0, y=1,
    inner_radius=0.3,
    outer_radius=0.8,
    start_angle=cumsum('angle', include_zero=True),
    end_angle=cumsum('angle'),
    line_color="white",
    fill_color='color',
    legend_field='Promotion',
    source=source,
    selection_color="firebrick",
    nonselection_fill_alpha=0.4
)

p.axis.visible = False
p.grid.grid_line_color = None
p.legend.location = "top_right"
p.legend.label_text_font_size = "10pt"
p.legend.click_policy = "hide"

show(p)
```

#### Insight #3

The number of transactions is slightly higher for non-promoted items (132 vs 120), reinforcing that non-promoted products dominate in volume, while promotions improve per-transaction performance.

#### Insight #4

This shows that promotions can enhance the average purchase value but do not necessarily increase total revenue, highlighting the need to balance promotion strategy with volume.

# Total Sales By Seasonality  
*Interactive pie chart showing the proportion of total sales in and out of season*

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import cumsum
from bokeh.io import output_notebook
import pandas as pd
import math

output_notebook()

seasonal_summary['Seasonal_Label'] = seasonal_summary['Seasonal'].map({
    'Yes': 'In Season',
    'No': 'Out of Season'
})

seasonal_summary['angle'] = seasonal_summary['Total_Sales'] / seasonal_summary['Total_Sales'].sum() * 2 * math.pi
seasonal_summary['color'] = ["#605046", "#A4968A", "#C7B299", "#E0D8C3"][:len(seasonal_summary)]

source = ColumnDataSource(seasonal_summary)

p = figure(
    height=400, width=400,
    title="Total Sales by Seasonality",
    toolbar_location=None,
    tools="hover",
    tooltips="@Seasonal_Label: @Total_Sales{0,0}"
)

p.wedge(
    x=0, y=1, radius=0.8,
    start_angle=cumsum('angle', include_zero=True),
    end_angle=cumsum('angle'),
    line_color="white",
    fill_color='color',
    legend_field='Seasonal_Label',
    source=source
)

p.axis.visible = False
p.grid.grid_line_color = None
p.legend.location = "top_right"
p.legend.label_text_font_size = "10pt"

show(p)
```

# Total Sales By Sub Category In & Out Of Season
*Interactive pie chart showing the proportion of total sales by sub-category in or out of season*

```python
from bokeh.plotting import figure, show
from bokeh.models import ColumnDataSource, HoverTool
from bokeh.transform import cumsum
from bokeh.io import output_notebook
import pandas as pd
import math

output_notebook()

seasonal_subcat_summary['Seasonal_Label'] = seasonal_subcat_summary['Seasonal'].map({
    'Yes': 'In Season',
    'No': 'Out of Season'
})

season_labels = seasonal_subcat_summary['Seasonal_Label'].unique()
colors = ["#605046", "#A4968A", "#C7B299", "#E0D8C3", "#B3A394", "#8C7B6B"]

for season in season_labels:
    data = seasonal_subcat_summary[seasonal_subcat_summary['Seasonal_Label'] == season].copy()
    data['angle'] = data['Total_Sales'] / data['Total_Sales'].sum() * 2 * math.pi
    data['color'] = colors[:len(data)]
    source = ColumnDataSource(data)

    p = figure(
        height=400, width=400,
        title=f"Total Sales by Sub-Category ({season})",
        toolbar_location=None,
        tools="hover",
        tooltips="@Sub_Category: @Total_Sales{0,0}"
    )

    p.wedge(
        x=0, y=1, radius=0.8,
        start_angle=cumsum('angle', include_zero=True),
        end_angle=cumsum('angle'),
        line_color="white", fill_color='color',
        legend_field='Sub_Category',
        source=source
    )

    p.axis.visible = False
    p.grid.grid_line_color = None
    p.legend.location = "top_right"
    p.legend.label_text_font_size = "10pt"

    show(p)
```

#### Insight #1

Total sales are slightly higher “In Season” (233,181) compared to out-of-season (226,392), showing that seasonal demand drives a modest increase in overall revenue.

#### Insight #2

The Jacket sub-category has almost equal sales in-season (111,868) and out-of-season (110,980), indicating steady demand regardless of season, unlike other categories.

#### Insight #3

Other categories, such as Sweater, show strong seasonal variation: in-season sales (52,235) are more than double out-of-season sales (20,270), showing sweaters are highly sensitive to seasonality.

#### Insight #4

Average sales per transaction remain fairly consistent across seasons, around 1,820-1,830, suggesting that while total sales fluctuate with season, the typical transaction value is stable.

#### Insight #5

These pie charts emphasize which sub-categories are more seasonal versus stable, informing inventory management and marketing strategies for seasonal promotions.

## Key Takeaways

### Product Placement and Sales
- **Aisle placements** lead to the highest total sales and transaction counts, showing that high-traffic areas drive volume.  
- **Front Of Store** placements generate fewer transactions but achieve the **highest average sales per transaction**, indicating strategic placement boosts per-sale revenue.  
- **End-Cap** placements perform moderately on both total and average sales.  

### Product Categories Driving Sales
- **Jackets** are the top-performing sub-category in total sales and number of transactions.  
- **Shirts and Sweaters** show moderate total sales with stable average sales.  
- **Pants** and **Other** have lower total sales, highlighting niche demand segments.  

### Pricing Impact
- Sales are concentrated in the mid-price ranges (51-100), which generate the highest total revenue.  
- Higher-priced items (200+) sell fewer units, but average sales per transaction remain relatively stable, showing consistent spending per customer.  

### Promotion Effectiveness
- Non-promoted items generate higher total sales, but promoted items maintain slightly higher average sales per transaction.  
- Promotions can increase transaction volume without drastically affecting revenue per sale, suggesting they are effective for boosting exposure.  

### Seasonality Effects
- “In season” periods slightly outperform “out of season” in total sales for most sub-categories.  
- **Average sales per transaction remain stable** across seasons, indicating that customer spending per transaction is consistent regardless of timing.  

### Strategic Implications
- Focus on **mid-range pricing** for volume while keeping premium options for higher per-transaction revenue.  
- Allocate **high-traffic placements** to top-selling categories like Jackets for maximum revenue impact.  
- Use promotions strategically to boost visibility and transaction counts.  
- Plan inventory and marketing campaigns around seasonality trends to maintain steady sales throughout the year.
